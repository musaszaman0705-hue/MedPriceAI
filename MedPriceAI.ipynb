```python

# STEP 2: Build the MedPrice AI Dataset

import pandas as pd

# Sample medication price data
# IMPORTANT: All prices below are fictional demo prices.

data = {
    "medication": [
        "Metformin",
        "Metformin",
        "Metformin",
        "Amoxicillin",
        "Amoxicillin",
        "Amoxicillin",
        "Atorvastatin",
        "Atorvastatin",
        "Atorvastatin",
        "Lisinopril",
        "Lisinopril",
        "Lisinopril",
        "Sertraline",
        "Sertraline",
        "Sertraline"
    ],
    "strength": [
        "500 mg", "500 mg", "500 mg",
        "500 mg", "500 mg", "500 mg",
        "20 mg", "20 mg", "20 mg",
        "10 mg", "10 mg", "10 mg",
        "50 mg", "50 mg", "50 mg"
    ],
    "quantity": [
        30, 30, 30, 21, 21, 21, 30, 30, 30,
        30, 30, 30, 30, 30, 30
    ],
    "pharmacy": [
        "Community Pharmacy",
        "City Pharmacy",
        "Neighborhood Drugs",
        "Community Pharmacy",
        "City Pharmacy",
        "Neighborhood Drugs",
        "Community Pharmacy",
        "City Pharmacy",
        "Neighborhood Drugs",
        "Community Pharmacy",
        "City Pharmacy",
        "Neighborhood Drugs",
        "Community Pharmacy",
        "City Pharmacy",
        "Neighborhood Drugs"
    ],
    "zip_code": [
        "77002", "77002", "77002",
        "77002", "77002", "77002",
        "77002", "77002", "77002",
        "77002", "77002", "77002",
        "77002", "77002", "77002"
    ],
    "price": [
        8.50, 6.25, 7.10,
        12.00, 9.50, 10.25,
        11.75, 8.00, 9.25,
        7.50, 5.99, 6.75,
        13.50, 10.00, 11.25
    ]
}

medication_df = pd.DataFrame(data)

# Display the dataset
print("MedPrice AI: Medication Price Dataset")
print("These are fictional demo prices, not current pharmacy quotes.")
print(f"\nNumber of records: {len(medication_df)}")

display(medication_df)
```

```python

# Check the dataset structure

print("Dataset information:")
medication_df.info()

print("\nAvailable medications:")
print(medication_df["medication"].unique())

print("\nAvailable pharmacies:")
print(medication_df["pharmacy"].unique())

print("\nPrice summary:")
print(medication_df["price"].describe())

```

```python

# STEP 3: Clean and Preprocess the Dataset

# Create a copy so the original dataset stays unchanged
cleaned_df = medication_df.copy()

# 1. Standardize medication names and text fields
cleaned_df["medication"] = (
    cleaned_df["medication"].str.strip().str.title()
)

cleaned_df["strength"] = cleaned_df["strength"].str.strip()
cleaned_df["pharmacy"] = cleaned_df["pharmacy"].str.strip()
cleaned_df["zip_code"] = cleaned_df["zip_code"].astype(str).str.strip()

# 2. Make sure prices are numeric
cleaned_df["price"] = pd.to_numeric(
    cleaned_df["price"], errors="coerce"
)

# 3. Remove records missing important information
required_columns = [
    "medication",
    "strength",
    "quantity",
    "pharmacy",
    "zip_code",
    "price"
]

cleaned_df = cleaned_df.dropna(subset=required_columns)

# 4. Remove invalid prices and quantities
cleaned_df = cleaned_df[
    (cleaned_df["price"] > 0) &
    (cleaned_df["quantity"] > 0)
].copy()

# 5. Remove duplicate records
cleaned_df = cleaned_df.drop_duplicates()

# 6. Reset row numbering
cleaned_df = cleaned_df.reset_index(drop=True)

print("Dataset cleaning completed!")
print(f"Original records: {len(medication_df)}")
print(f"Clean records: {len(cleaned_df)}")
print(f"Records removed: {len(medication_df) - len(cleaned_df)}")

display(cleaned_df)
```

```python

# Check the quality of the cleaned dataset

print("Missing values in each column:")
print(cleaned_df.isnull().sum())

print("\nDuplicate records remaining:")
print(cleaned_df.duplicated().sum())

print("\nInvalid prices remaining:")
print((cleaned_df["price"] <= 0).sum())

print("\nCleaned dataset summary:")
cleaned_df.info()

```

```python

# STEP 4: Build the MedPrice AI Comparison Engine

def compare_prices(medication, zip_code="77002", quantity=None):
    """
    Find and compare demo prices for a medication.
    Prices in this prototype are fictional.
    """

    # Clean the user's search input
    medication = medication.strip().title()
    zip_code = str(zip_code).strip()

    # Start with the cleaned dataset
    results = cleaned_df[
        (cleaned_df["medication"] == medication) &
        (cleaned_df["zip_code"] == zip_code)
    ].copy()

    # Filter by quantity if the user provides one
    if quantity is not None:
        results = results[
            results["quantity"] == int(quantity)
        ]

    # Handle medications that aren't found
    if results.empty:
        print("No matching medication records found.")
        print("Try another medication or ZIP code.")
        return None

    # Sort pharmacies from cheapest to most expensive
    results = results.sort_values("price").reset_index(drop=True)

    # Calculate the price difference
    lowest_price = results["price"].min()
    highest_price = results["price"].max()
    potential_difference = highest_price - lowest_price

    print(f"Medication: {medication}")
    print(f"ZIP code: {zip_code}")
    print("\nFictional demo prices — not live pharmacy quotes.\n")

    display(
        results[["medication", "strength", "quantity",
                 "pharmacy", "zip_code", "price"]]
    )

    print(f"\nLowest listed demo price: ${lowest_price:.2f}")
    print(f"Highest listed demo price: ${highest_price:.2f}")
    print(f"Difference between prices: ${potential_difference:.2f}")

    return results
```

```python

# Test 1: Compare prices for Metformin
metformin_results = compare_prices(
    medication="Metformin",
    zip_code="77002",
    quantity=30
)
```
